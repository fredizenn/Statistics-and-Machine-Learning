### 3. Cross-Model Discussion Points

| Question | Notes |
|---|---|
| Whose model scored highest on Accuracy? | RF/GB, at 0.779 — the ensemble edge held up here even though it didn't win on every metric. |
| Whose model scored highest on Recall? | Logistic Regression, at 0.814, well ahead of the others. |
| Are the "best accuracy" and "best recall" the same student? | No — RF/GB wins accuracy, Logistic Regression wins recall. The two priorities pulled in different directions. |
| Which model would you trust most for a real clinical screening tool, and why (Precision vs. Recall priority)? | Logistic Regression. In a screening context, missing an actual diabetic case (false negative) is worse than flagging a healthy person for a follow-up test (false positive), so we'd rather optimize recall even at the cost of some precision — and LR's 0.814 recall is the clear standout. |
| Which features showed up as "most important" across Decision Tree / Random Forest / Gradient Boosting? Do they agree? | Yes, strong agreement. **Glucose** was the #1 feature in all three models by a wide margin, and **BMI** was #2 in all three as well (Decision Tree: Glucose 0.27, BMI 0.15; Random Forest: Glucose 0.27, BMI 0.16; Gradient Boosting: Glucose 0.41, BMI 0.19). Logistic Regression's coefficients back this up too — Glucose (1.14) and BMI (0.71) had the largest coefficients of any feature. After that the models diverge a bit: DiabetesPedigreeFunction and Age trade off for 3rd/4th place depending on the model, and SkinThickness/BloodPressure consistently rank near the bottom across all three. |
| Did scaling matter a lot (KNN, Logistic Regression) vs. not at all (Trees, RF, GB)? | Yes. KNN and Logistic Regression both rely on distances/gradients over feature magnitudes, so unscaled features skew results — scaling changed KNN's accuracy noticeably (0.67 → 0.75, see Section 4). Tree-based models split on thresholds per feature independently, so scaling had no real effect there. |
| Which model was easiest to explain to a non-ML person? | The Decision Tree — you can literally draw the flowchart and walk someone through the yes/no splits without touching any math. |
| Which visualization (across all 5 exercises) was most convincing for explaining *why* a model performed as it did? | The Logistic Regression **precision-recall vs. threshold** line chart. It shows precision and recall moving in opposite directions as the threshold slides from 0.05 to 0.95, which makes the accuracy/recall trade-off we kept discussing immediately obvious — far more convincing than reading the numbers off a table. The feature importance bar charts were a close second since they made the Glucose/BMI dominance visually undeniable across all three tree-based models. |

### 6. Group Summary Paragraph

*(5 sentences, written together as a group wrapping up week 3's key takeaways )*

> Across the four classification exercises, Random Forest and Gradient Boosting produced the strongest overall accuracy, while plain Logistic Regression stood out for recall, making it the model we'd lean on if catching every possible diabetic case mattered more than avoiding false alarms. Glucose and BMI came out as the two most important predictors across every single model we tried — Decision Tree, Random Forest, Gradient Boosting, and even the Logistic Regression coefficients agreed on that ranking, which gave us real confidence in those two features. Scaling made a real difference for KNN and Logistic Regression but barely moved the needle for the tree-based models, which lines up with how each algorithm actually uses the input features — distances and gradients versus simple threshold splits. The Decision Tree was the easiest model to walk a non-technical person through, since the tree structure basically explains itself without any extra translation. Overall, tuning hyperparameters — threshold, depth, k, n_estimators — shifted the precision/recall trade-off far more than it changed raw accuracy, which was probably the biggest takeaway of the week.
