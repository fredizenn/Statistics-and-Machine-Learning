```python
from scipy import stats
group1 = df[df['class'] == 1]['Alcohol']
group2 = df[df['class'] == 2]['Alcohol']
stats.ttest_ind(group1, group2)
```

```python
group1 = df[df['class'] == 1]['Alcohol']
group2 = df[df['class'] == 2]['Alcohol']
group3 = df[df['class'] == 3]['Alcohol']
stats.f_oneway(group1, group2, group3)
```
