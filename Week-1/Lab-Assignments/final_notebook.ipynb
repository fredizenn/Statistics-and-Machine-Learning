>
> **3.** Pregnancies is discrete. Having only 17 distinct values does not by itself determine the type — it is the nature of the underlying quantity that matters. ### ✍️ Interpretation: Task 3

1. Which feature was most affected (largest mean shift) by removing zeros? Why does removing zeros increase the mean?
2. Did the median shift by more or less than the mean? Relate this to what you know about median robustness.
3. If you were building a predictive model, would you drop rows with NaN or impute them? What are the trade-offs?

> **1.** Insulin experienced the largest mean shift after replacing zeros with NaN — roughly from 79.8 to 155.5, an increase of about 75.7 units. Zeros artificially drag the mean downward because the mean sums all values and divides by the total count. When a value of 0 is included in that sum, it contributes nothing to the numerator while still incrementing the denominator, pulling the average below the true center of the real measurements.
>
> **2.** The median shifted by less than the mean. This is a direct illustration of median robustness: the median only depends on the rank ordering of values, not their actual magnitudes. A string of zeros at the bottom of the sorted list simply shifts the positional midpoint somewhat, but once those zeros are removed and the remaining values are re-ranked, the median settles close to the genuine center. The mean has no such protection — every single value, including the implausible zeros, enters the formula with equal weight.
>
> **3.** We would generally impute rather than drop. Dropping rows with NaN preserves the assumption that the remaining data is unbiased, but with a column like Insulin where ~48.7% of rows have missing values, dropping them would cut the dataset nearly in half and likely bias the remaining sample toward patients who had insulin measured — possibly a systematically different subgroup. Imputation keeps more data, though it introduces assumptions: mean/median imputation is simple but artificially reduces variance and can distort correlations; model-based imputation (e.g. KNN or iterative imputer) is more principled but adds complexity and risk of overfitting. The best choice depends on the missingness mechanism and the downstream model.

> **1.** Glucose shows the largest mean difference between groups: roughly 141.3 mg/dL for diabetic patients versus 110.0 mg/dL for non-diabetic patients — a gap of about 31 mg/dL. Clinically this lines up with what we know: sustained high blood glucose is both a defining criterion and a consequence of type 2 diabetes, so a strong mean difference is exactly what we would expect. >
> **2.** Insulin is the most positively skewed feature (skewness ≈ 2.17). The long tail stretches towards high values — a small number of patients have very high serum insulin readings, while the bulk of the population clusters at lower values. In practice this arises because insulin resistance (common in pre-diabetic and diabetic patients) causes the pancreas to over-secrete insulin, producing a handful of very high readings that pull the right tail out. The large number of encoded zeros also contributes by creating a spike near zero.
>
> **3.** The mean changed substantially more than the median after adding five outlier values of 900. The mean formula is μ = (1/n) Σxᵢ — every data point, no matter how extreme, enters the sum with equal weight, so a single value of 900 contributes 900/n to the total, which can shift the mean significantly. The median formula, by contrast, requires only that we find the middle-ranked value after sorting. Adding five large outliers at the top of the sorted list does not change which value sits in the middle position — the median barely moves.

# Module C: Fundamentals of Probability
*Day 2 | Tasks 6, 7*
> Copy & Paste from "module_c.ipynb"

```python
# ── Task 6 Setup ──────────────────────────────────────────────────────────────
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (9, 4)

URL = 'https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv'
df = pd.read_csv(URL)
df['DiabetesStatus'] = df['Outcome'].map({0: 'Non-Diabetic', 1: 'Diabetic'})

HIGH_GLUCOSE_THRESHOLD = 140
HIGH_BMI_THRESHOLD     = 30
n = len(df)
print(f'Dataset loaded: {n} patients')

# Base rate
p_diabetic     = df['Outcome'].mean()
p_non_diabetic = 1 - p_diabetic
print(f'\nP(Diabetic)     = {p_diabetic:.4f}  ({p_diabetic*100:.1f}%)')
print(f'P(Non-Diabetic) = {p_non_diabetic:.4f}  ({p_non_diabetic*100:.1f}%)')

# P(A) and P(B)
p_A = (df['Glucose'] > HIGH_GLUCOSE_THRESHOLD).mean()
p_B = (df['BMI']     > HIGH_BMI_THRESHOLD).mean()
print(f'\nP(Glucose > 140) = {p_A:.4f}  ({p_A*100:.1f}%)')
print(f'P(BMI > 30)       = {p_B:.4f}  ({p_B*100:.1f}%)')

# Addition rule
p_A_and_B    = ((df['Glucose'] > HIGH_GLUCOSE_THRESHOLD) & (df['BMI'] > HIGH_BMI_THRESHOLD)).mean()
p_A_or_B     = p_A + p_B - p_A_and_B
p_A_or_B_dir = ((df['Glucose'] > HIGH_GLUCOSE_THRESHOLD) | (df['BMI'] > HIGH_BMI_THRESHOLD)).mean()
print(f'\nP(A ∩ B)             = {p_A_and_B:.4f}')
print(f'P(A ∪ B) [formula]   = {p_A_or_B:.4f}')
print(f'P(A ∪ B) [direct]    = {p_A_or_B_dir:.4f}')
print(f'Match: {round(p_A_or_B, 10) == round(p_A_or_B_dir, 10)}')

# Independent multiplication rule
p_both_high_glucose = p_A * p_A
print(f'\nP(both patients high glucose) = {p_both_high_glucose:.4f} ({p_both_high_glucose*100:.2f}%)')
print('Interpretation: Uncommon — roughly 1 in 16 pairs of randomly selected patients.')

# Conditional probability
high_glucose_patients = df[df['Glucose'] > HIGH_GLUCOSE_THRESHOLD]
p_diab_given_high_gluc = (high_glucose_patients['Outcome'] == 1).mean()
lift = p_diab_given_high_gluc / p_diabetic
print(f'\nP(Diabetic)                  = {p_diabetic:.4f}')
print(f'P(Diabetic | Glucose > 140)  = {p_diab_given_high_gluc:.4f}')
print(f'Lift over base rate          = {lift:.2f}x')

# Grouped bar chart
p_high_gluc_diab   = (df[df['Glucose'] > HIGH_GLUCOSE_THRESHOLD]['Outcome'] == 1).mean()
p_normal_gluc_diab = (df[df['Glucose'] <= HIGH_GLUCOSE_THRESHOLD]['Outcome'] == 1).mean()
p_high_bmi_diab    = (df[df['BMI'] > HIGH_BMI_THRESHOLD]['Outcome'] == 1).mean()
p_normal_bmi_diab  = (df[df['BMI'] <= HIGH_BMI_THRESHOLD]['Outcome'] == 1).mean()

plot_df = pd.DataFrame({
    'High (Above Threshold)':  [p_high_gluc_diab,   p_high_bmi_diab],
    'Normal (Below Threshold)': [p_normal_gluc_diab, p_normal_bmi_diab]
}, index=['Glucose > 140', 'BMI > 30'])

fig, ax = plt.subplots(figsize=(9, 6))
plot_df.plot(kind='bar', ax=ax, color=['#e74c3c', '#2ecc71'], width=0.5, edgecolor='white')
for container in ax.containers:
    for bar in container:
        h = bar.get_height()
        ax.text(bar.get_x() + bar.get_width() / 2, h + 0.01,
                f'{h*100:.1f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')
ax.axhline(y=p_diabetic, color='gray', linestyle='--', lw=1.2,
           label=f'Baseline P(Diabetic) = {p_diabetic*100:.1f}%')
ax.set_title('P(Diabetic | Condition) — Glucose vs BMI', fontsize=14, fontweight='bold')
ax.set_ylabel('Probability of Being Diabetic')
ax.set_xlabel('Risk Factor')
ax.set_ylim(0, 1.0)
ax.set_xticklabels(plot_df.index, rotation=0, fontsize=12)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y*100:.0f}%'))
ax.legend(fontsize=10)
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# ── Task 7: Simulation vs Formula ────────────────────────────────────────────
np.random.seed(42)
HIGH_GLUCOSE_THRESHOLD = 140

# Independent multiplication rule simulation
p_high    = (df['Glucose'] > HIGH_GLUCOSE_THRESHOLD).mean()
draws     = np.random.choice(df['Glucose'].values, size=(10_000, 2), replace=True)
sim_both  = np.all(draws > HIGH_GLUCOSE_THRESHOLD, axis=1).mean()
analytic  = p_high ** 2
print(f'\n── Independent Simulation ──')
print(f'Analytical P(both high glucose) = {analytic:.4f}')
print(f'Simulated  P(both high glucose) = {sim_both:.4f}')
print(f'Absolute difference             = {abs(analytic - sim_both):.4f}')

# Dependent multiplication rule simulation
np.random.seed(42)
n_diabetic = (df['Outcome'] == 1).sum()
p_first    = n_diabetic / len(df)
p_second   = (n_diabetic - 1) / (len(df) - 1)
analytic_dep = p_first * p_second
sim_dep = np.array([np.random.choice(df['Outcome'].values, size=2, replace=False)
                    for _ in range(10_000)])
sim_dep_prop = np.all(sim_dep == 1, axis=1).mean()
print(f'\n── Dependent Simulation ──')
print(f'Analytical P(both diabetic, no replacement) = {analytic_dep:.4f}')
print(f'Simulated  P(both diabetic, no replacement) = {sim_dep_prop:.4f}')
print(f'Absolute difference                          = {abs(analytic_dep - sim_dep_prop):.4f}')
print(f'Difference vs independent case (p²)         = {abs(analytic_dep - analytic):.6f}')

# Convergence plot
np.random.seed(42)
true_prob    = p_high ** 2
trial_sizes  = [10, 50, 100, 500, 1000, 5000, 10000]
sim_probs    = []
for t in trial_sizes:
    s = np.random.choice(df['Glucose'].values, size=(t, 2), replace=True)
    sim_probs.append(np.all(s > HIGH_GLUCOSE_THRESHOLD, axis=1).mean())

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(trial_sizes, sim_probs, marker='o', label='Simulated Probability')
ax.axhline(y=true_prob, color='red', linestyle='--', label=f'True P = {true_prob:.4f}')
ax.set_xscale('log')
ax.set_xlabel('Number of Simulation Trials')
ax.set_ylabel('Estimated Probability')
ax.set_title('Convergence of Simulated Probability to Analytical Result')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# ── Task 8 Setup ──────────────────────────────────────────────────────────────
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (14, 4)
np.random.seed(42)

URL = 'https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv'
df = pd.read_csv(URL)
df['DiabetesStatus'] = df['Outcome'].map({0: 'Non-Diabetic', 1: 'Diabetic'})
N = 768
print(f'Ready (N = {N})')

# Normal distribution: synthetic vs real Glucose
mu, sigma = df['Glucose'].mean(), df['Glucose'].std()
normal_data = np.random.normal(loc=mu, scale=sigma, size=N)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(normal_data, bins=30, color='steelblue', edgecolor='white', density=True)
x = np.linspace(mu - 3.5 * sigma, mu + 3.5 * sigma, 300)
axes[0].plot(x, stats.norm.pdf(x, mu, sigma), 'r-', lw=2, label='Normal PDF')
axes[0].set_title(f'Synthetic Normal  μ={mu:.1f}, σ={sigma:.1f}\nskewness={stats.skew(normal_data):.2f}')
axes[0].set_xlabel('Value'); axes[0].legend()
mu_g, sigma_g = df['Glucose'].mean(), df['Glucose'].std()
axes[1].hist(df['Glucose'], bins=30, color='coral', edgecolor='white', density=True)
x_g = np.linspace(mu_g - 3.5 * sigma_g, mu_g + 3.5 * sigma_g, 300)
axes[1].plot(x_g, stats.norm.pdf(x_g, mu_g, sigma_g), 'r-', lw=2, label='Normal PDF fit')
axes[1].set_title(f'Real Glucose  μ={mu_g:.1f}, σ={sigma_g:.1f}\nskewness={stats.skew(df["Glucose"]):.2f}')
axes[1].set_xlabel('Glucose (mg/dL)'); axes[1].legend()
plt.suptitle('Normal Distribution: Synthetic vs Real', fontsize=12, fontweight='bold')
plt.tight_layout(); plt.show()

# Uniform distribution: synthetic vs real BloodPressure
np.random.seed(42)
low, high = df['BloodPressure'].min(), df['BloodPressure'].max()
uniform_data = np.random.uniform(low=low, high=high, size=N)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(uniform_data, bins=30, color='steelblue', edgecolor='white', density=True)
x_u = np.linspace(low, high, 300)
axes[0].plot(x_u, stats.uniform.pdf(x_u, loc=low, scale=high - low), 'r-', lw=2, label='Uniform PDF')
axes[0].set_title(f'Synthetic Uniform  [{low:.0f}, {high:.0f}]\nskewness={stats.skew(uniform_data):.2f}')
axes[0].set_xlabel('Value'); axes[0].legend()
mu_bp, sigma_bp = df['BloodPressure'].mean(), df['BloodPressure'].std()
axes[1].hist(df['BloodPressure'], bins=30, color='coral', edgecolor='white', density=True)
x_bp = np.linspace(mu_bp - 3.5 * sigma_bp, mu_bp + 3.5 * sigma_bp, 300)
axes[1].plot(x_bp, stats.norm.pdf(x_bp, mu_bp, sigma_bp), 'r-', lw=2, label='Normal PDF fit')
axes[1].set_title(f'Real BloodPressure  μ={mu_bp:.1f}, σ={sigma_bp:.1f}\nskewness={stats.skew(df["BloodPressure"]):.2f}')
axes[1].set_xlabel('BloodPressure (mm Hg)'); axes[1].legend()
plt.suptitle('Uniform Distribution: Synthetic vs Real', fontsize=12, fontweight='bold')
plt.tight_layout(); plt.show()
print('BloodPressure does not look uniform — it peaks around 60–80 mm Hg with a left tail (skewness ≈ -1.84).')

# Binomial distribution
np.random.seed(42)
n_b, p_b = 10, 0.65
binom_data = np.random.binomial(n=n_b, p=p_b, size=5000)
values_b, counts_b = np.unique(binom_data, return_counts=True)
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(values_b, counts_b / counts_b.sum(), color='steelblue', edgecolor='white')
ax.axvline(n_b * p_b, color='red', linestyle='--', lw=2, label=f'E[X] = n·p = {n_b*p_b:.1f}')
ax.set_title(f'Binomial (n={n_b}, p={p_b})  skewness={stats.skew(binom_data):.2f}')
ax.set_xlabel('Patients Correctly Flagged (out of 10)'); ax.set_ylabel('Relative Frequency')
ax.set_xticks(range(0, 11)); ax.legend()
plt.tight_layout(); plt.show()
print(f'Observed mean: {binom_data.mean():.4f}  |  Theoretical E[X] = n·p = {n_b*p_b:.4f}')
print(f'Observed var:  {binom_data.var():.4f}   |  Theoretical Var  = n·p·(1-p) = {n_b*p_b*(1-p_b):.4f}')

# Poisson distribution
np.random.seed(42)
lam = 4
poisson_data = np.random.poisson(lam=lam, size=1000)
values_p, counts_p = np.unique(poisson_data, return_counts=True)
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(values_p, counts_p / counts_p.sum(), color='steelblue', edgecolor='white')
ax.axvline(lam, color='red', linestyle='--', lw=2, label=f'λ = {lam}')
ax.set_title(f'Poisson (λ={lam})  skewness={stats.skew(poisson_data):.2f}')
ax.set_xlabel('Daily Referrals'); ax.set_ylabel('Relative Frequency')
ax.set_xticks(range(0, poisson_data.max() + 1)); ax.legend()
plt.tight_layout(); plt.show()
print(f'Simulated mean: {poisson_data.mean():.4f}  |  Theoretical λ = {lam}')
print(f'Simulated var:  {poisson_data.var():.4f}   |  Theoretical λ = {lam}')

# 2×2 summary grid
np.random.seed(42)
fig, axes = plt.subplots(2, 2, figsize=(13, 8))
mu, sigma = df['Glucose'].mean(), df['Glucose'].std()
normal_data = np.random.normal(loc=mu, scale=sigma, size=768)
axes[0, 0].hist(normal_data, bins=30, color='steelblue', edgecolor='white', density=True)
x = np.linspace(mu - 3.5 * sigma, mu + 3.5 * sigma, 300)
axes[0, 0].plot(x, stats.norm.pdf(x, mu, sigma), 'r-', lw=2)
axes[0, 0].set_title(f'Normal  (μ={mu:.1f}, σ={sigma:.1f})\nskewness={stats.skew(normal_data):.2f}')
axes[0, 0].set_xlabel('Glucose (mg/dL)')
low, high = df['BloodPressure'].min(), df['BloodPressure'].max()
uniform_data = np.random.uniform(low=low, high=high, size=768)
axes[0, 1].hist(uniform_data, bins=30, color='steelblue', edgecolor='white', density=True)
x_u = np.linspace(low, high, 300)
axes[0, 1].plot(x_u, stats.uniform.pdf(x_u, loc=low, scale=high - low), 'r-', lw=2)
axes[0, 1].set_title(f'Uniform  (low={low:.0f}, high={high:.0f})\nskewness={stats.skew(uniform_data):.2f}')
axes[0, 1].set_xlabel('BloodPressure (mm Hg)')
binom_data = np.random.binomial(n=10, p=0.65, size=5000)
v_b, c_b = np.unique(binom_data, return_counts=True)
axes[1, 0].bar(v_b, c_b / c_b.sum(), color='steelblue', edgecolor='white')
axes[1, 0].axvline(10 * 0.65, color='red', linestyle='--', lw=2, label='E[X]=6.5')
axes[1, 0].set_title(f'Binomial  (n=10, p=0.65)\nskewness={stats.skew(binom_data):.2f}')
axes[1, 0].set_xlabel('Patients Correctly Flagged'); axes[1, 0].set_xticks(range(0, 11)); axes[1, 0].legend()
poisson_data = np.random.poisson(lam=4, size=1000)
v_p, c_p = np.unique(poisson_data, return_counts=True)
axes[1, 1].bar(v_p, c_p / c_p.sum(), color='steelblue', edgecolor='white')
axes[1, 1].axvline(4, color='red', linestyle='--', lw=2, label='λ=4')
axes[1, 1].set_title(f'Poisson  (λ=4)\nskewness={stats.skew(poisson_data):.2f}')
axes[1, 1].set_xlabel('Daily Referrals'); axes[1, 1].set_xticks(range(0, poisson_data.max() + 1)); axes[1, 1].legend()
plt.suptitle('The Four Key Distributions', fontsize=13, fontweight='bold')
plt.tight_layout(); plt.show()

# ── Task 9: Skewness ──────────────────────────────────────────────────────────
NUMERIC_COLS = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
                'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']

skewness = df[NUMERIC_COLS].skew().sort_values(ascending=False).round(3)
fig, ax = plt.subplots(figsize=(9, 4))
colors = ['tomato' if s > 0 else 'steelblue' for s in skewness]
skewness.plot(kind='bar', ax=ax, color=colors, edgecolor='white')
ax.axhline(0, color='black', linewidth=0.8, linestyle='--')
ax.axhline( 0.5, color='orange', linewidth=0.8, linestyle=':', label='+/-0.5 threshold')
ax.axhline(-0.5, color='orange', linewidth=0.8, linestyle=':')
ax.set_title('Skewness of All Numeric Features (red = positive, blue = negative)')
ax.set_ylabel('Skewness'); ax.tick_params(axis='x', rotation=30); ax.legend()
for i, v in enumerate(skewness):
    ax.text(i, v + (0.05 if v >= 0 else -0.15), f'{v:.2f}', ha='center', fontsize=8)
plt.tight_layout(); plt.show()
print('\nSkewness values:'); print(skewness)

# Mean–Median gap table
results = []
for col in NUMERIC_COLS:
    mean   = df[col].mean()
    median = df[col].median()
    results.append([col, mean, median, mean - median, df[col].skew()])
gap_table = pd.DataFrame(results, columns=['Feature', 'Mean', 'Median', 'Gap (Mean−Median)', 'Skewness'])
gap_table = gap_table.sort_values(by='Gap (Mean−Median)', key=abs, ascending=False).round(4)
print('\nMean–Median gap table:'); print(gap_table)

# Insulin anatomy plot
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(df['Insulin'], bins=30, color='steelblue', edgecolor='white')
mean_ins   = df['Insulin'].mean()
median_ins = df['Insulin'].median()
mode_ins   = df['Insulin'].mode()[0]
ax.axvline(mean_ins,   color='red',    linestyle='-',  lw=2, label=f'Mean = {mean_ins:.2f}')
ax.axvline(median_ins, color='green',  linestyle='--', lw=2, label=f'Median = {median_ins:.2f}')
ax.axvline(mode_ins,   color='purple', linestyle=':',  lw=2, label=f'Mode = {mode_ins:.2f}')
ax.annotate('Long right tail (positive skew)', xy=(300, 20), xytext=(150, 40),
            arrowprops=dict(arrowstyle='->'))
ax.set_title(f'Insulin Distribution  (Skewness = {df["Insulin"].skew():.2f})')
ax.set_xlabel('Insulin (mu U/ml)'); ax.set_ylabel('Frequency'); ax.legend()
plt.tight_layout(); plt.show()

# Log transformation of Insulin
insulin_clean = df['Insulin'].replace(0, np.nan).dropna()
insulin_log   = np.log1p(insulin_clean)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(insulin_clean, bins=30, kde=True, ax=axes[0], color='steelblue')
axes[0].axvline(insulin_clean.mean(),   color='red',   linestyle='--', label=f'Mean={insulin_clean.mean():.1f}')
axes[0].axvline(insulin_clean.median(), color='green', linestyle='--', label=f'Median={insulin_clean.median():.1f}')
axes[0].set_title(f'Original Insulin  (skewness={insulin_clean.skew():.2f})'); axes[0].legend()
sns.histplot(insulin_log, bins=30, kde=True, ax=axes[1], color='coral')
axes[1].axvline(insulin_log.mean(),   color='red',   linestyle='--', label=f'Mean={insulin_log.mean():.2f}')
axes[1].axvline(insulin_log.median(), color='green', linestyle='--', label=f'Median={insulin_log.median():.2f}')
axes[1].set_title(f'Log-Transformed Insulin  (skewness={insulin_log.skew():.2f})'); axes[1].legend()
plt.suptitle('Insulin: Before vs After Log Transformation', fontsize=12, fontweight='bold')
plt.tight_layout(); plt.show()
log_comparison = pd.DataFrame({
    'Version':  ['Original', 'Log-Transformed'],
    'Mean':     [insulin_clean.mean(), insulin_log.mean()],
    'Median':   [insulin_clean.median(), insulin_log.median()],
    'Skewness': [insulin_clean.skew(), insulin_log.skew()]
}).round(3)
print('\nLog transformation comparison:'); print(log_comparison)

# Distribution matching
distribution_match = {
    'Glucose':                  {'Closest Distribution': 'Normal',                    'Reason': 'Near-symmetric bell shape, mild right skew (0.17)'},
    'BloodPressure':            {'Closest Distribution': 'Normal',                    'Reason': 'Approximately symmetric, slightly left-skewed (-1.84) due to zeros'},
    'Insulin':                  {'Closest Distribution': 'Right-skewed (not standard)', 'Reason': 'Strong long right tail, skewness = 2.17'},
    'BMI':                      {'Closest Distribution': 'Normal',                    'Reason': 'Near-symmetric, skewness = 0.60'},
    'DiabetesPedigreeFunction': {'Closest Distribution': 'Right-skewed (not standard)', 'Reason': 'Many small values with long positive tail, skewness = 1.92'},
    'Age':                      {'Closest Distribution': 'Poisson',                   'Reason': 'Discrete-like, right-skewed count data, skewness = 1.13'},
    'Pregnancies':              {'Closest Distribution': 'Poisson',                   'Reason': 'Count data with right skew, skewness = 0.90'},
}
dist_df = pd.DataFrame(distribution_match).T
print('\nDistribution matching:'); print(dist_df)
```

Overdispersion can arise from clustering (patients in the same clinic influence each other's referral patterns), hidden heterogeneity in the underlying rate λ, or excess zero counts. A Negative Binomial distribution is typically used instead when overdispersion is present, as it adds a parameter to model the extra variance.
>
> **2.** The key practical distinction is whether there is a fixed, known number of trials. Use the **Binomial** when you have a fixed number of independent trials each with a fixed success probability — for example, if we screen exactly 10 diabetic patients with a test that correctly flags each patient with probability 0.65, the number correctly flagged follows Binomial(10, 0.65). Use the **Poisson** when you are counting events occurring in a continuous interval (time, area) at a constant rate, without a fixed upper bound — for example, the number of new diabetes referrals arriving at a clinic per day at an average rate of λ = 4.
>
> **3.** The Normal distribution has a skewness closest to zero (≈ 0.00–0.05 due to random sampling noise). Zero skewness means the distribution is perfectly symmetric: the mean, median, and mode all coincide at the same central point. This is the defining property of a symmetric distribution — the left and right tails are mirror images of each other, so there is no direction in which the average is pulled away from the middle value.

### ✍️ Interpretation: Task 9

1. From your mean–median gap table: does the sign of the gap always match the sign of the skewness? Explain mathematically why a positive skew pulls the mean above the median.
2. After the log transformation of Insulin: by how much did the skewness decrease? Did the transformation make the mean and median closer together? Why is this useful before applying some machine learning models?
3. You matched features to theoretical distributions. Which feature was hardest to classify, and why? What additional information or test would help you make a more rigorous decision?

> **1.** Yes, in all cases the sign of the gap (mean − median) matches the sign of the skewness. A positive skew means there is a long right tail — a small number of very large values. The skewness formula raises deviations from the mean to the third power, (xᵢ − μ)³, which preserves the sign: large positive deviations from the right tail dominate and push the overall sum positive. Because the mean is the sum of all values divided by n, those large right-tail values pull the mean upward. The median, being determined by rank rather than magnitude, does not move toward the outliers, so it stays closer to the bulk of the data. The result is mean > median whenever skewness > 0.
>
> **2.** Skewness dropped from approximately 2.17 (original Insulin, zeros removed) to about −0.09 after applying log1p — a reduction of roughly 2.26 units. The mean and median also converged substantially: before the transformation they were separated by about 30 units (155.5 vs 125.0); after transformation the log-scale mean and median are almost identical (≈ 4.82 vs 4.84). This is useful because several ML algorithms — linear regression, logistic regression, LDA, PCA — either assume normally distributed inputs or are sensitive to the scale and symmetry of features. A highly skewed feature can dominate distance-based or variance-based calculations; normalising it via log transformation brings it closer to the symmetric regime these models were designed for.
>
> **3.** BMI was the hardest to classify. Its histogram looks roughly bell-shaped and its skewness (≈ 0.60) is close to but slightly above the ±0.5 symmetry threshold, making it ambiguous — it could reasonably be called approximately Normal or mildly right-skewed. To make a more rigorous decision we could use a Q-Q plot (quantile-quantile plot against a theoretical Normal), which would make any systematic deviation from the diagonal immediately visible, or run a formal normality test such as Shapiro-Wilk or Kolmogorov-Smirnov, both of which provide a p-value for the null hypothesis that the data follows a Normal distribution.
