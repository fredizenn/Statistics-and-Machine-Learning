```python
# Code for Tasks 9

# TODO: Apply a log transformation to reduce skewness in Insulin.
#
# Insulin contains zeros (missing values encoded as 0).
# Replace zeros with NaN first, then apply np.log1p() which computes log(1 + x)
# and is safe for values >= 0.
#
# Step 1: Create insulin_clean by replacing zeros with NaN, then dropping NaN.
# Step 2: Apply log1p to get insulin_log.
# Step 3: Plot side-by-side histograms: original (clean, zeros removed) | log-transformed.
#         Show mean, median, and skewness on each panel.
# Step 4: Print a comparison table:
#         | Version      | Mean   | Median | Skewness |
#         | Original     | ...    | ...    | ...      |
#         | Log-Transformed | ... | ...    | ...      |

# YOUR CODE HERE

insulin_clean = df['Insulin'].replace(0, np.nan).dropna()

insulin_log = np.log1p(insulin_clean)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(insulin_clean, bins=30, kde=True, ax=axes[0])
axes[0].set_title('Original Insulin')

axes[0].axvline(insulin_clean.mean(),
                color='red',
                linestyle='--',
                label=f"Mean={insulin_clean.mean():.1f}")

axes[0].axvline(insulin_clean.median(),
                color='green',
                linestyle='--',
                label=f"Median={insulin_clean.median():.1f}")

axes[0].legend()

sns.histplot(insulin_log, bins=30, kde=True, ax=axes[1])
axes[1].set_title('Log-Transformed Insulin')

axes[1].axvline(insulin_log.mean(),
                color='red',
                linestyle='--',
                label=f"Mean={insulin_log.mean():.2f}")

axes[1].axvline(insulin_log.median(),
                color='green',
                linestyle='--',
                label=f"Median={insulin_log.median():.2f}")

axes[1].legend()

plt.tight_layout()
plt.show()

comparison = pd.DataFrame({
    'Version': ['Original', 'Log-Transformed'],
    'Mean': [insulin_clean.mean(), insulin_log.mean()],
    'Median': [insulin_clean.median(), insulin_log.median()],
    'Skewness': [insulin_clean.skew(), insulin_log.skew()]
})

print(comparison.round(3))


# TODO: Match each real feature to its closest theoretical distribution.
#
# Complete the dictionary below by replacing each '???' with one of:
# 'Normal', 'Uniform', 'Binomial', 'Poisson', 'Right-skewed (not standard)', 'Left-skewed (not standard)'
#
# Use the skewness values and histogram shapes from the plots above to decide.
# Then print the dictionary as a formatted table.

distribution_match = {
    'Glucose':                  {'Closest distribution': '???', 'Reason': '???'},
    'BloodPressure':            {'Closest distribution': '???', 'Reason': '???'},
    'Insulin':                  {'Closest distribution': '???', 'Reason': '???'},
    'BMI':                      {'Closest distribution': '???', 'Reason': '???'},
    'DiabetesPedigreeFunction': {'Closest distribution': '???', 'Reason': '???'},
    'Age':                      {'Closest distribution': '???', 'Reason': '???'},
    'Pregnancies':              {'Closest distribution': '???', 'Reason': '???'},
}

# YOUR CODE HERE (print as a DataFrame)

distribution_match = {
    'Glucose': {
        'Closest distribution': 'Normal',
        'Reason': 'Roughly bell-shaped with mild right skew'
    },
    'BloodPressure': {
        'Closest distribution': 'Normal',
        'Reason': 'Approximately symmetric around the center'
    },
    'Insulin': {
        'Closest distribution': 'Right-skewed (not standard)',
        'Reason': 'Strong long right tail and extreme values'
    },
    'BMI': {
        'Closest distribution': 'Normal',
        'Reason': 'Near-symmetric with moderate concentration around mean'
    },
    'DiabetesPedigreeFunction': {
        'Closest distribution': 'Right-skewed (not standard)',
        'Reason': 'Many small values and a long positive tail'
    },
    'Age': {
        'Closest distribution': 'Poisson',
        'Reason': 'Discrete-like positive values with moderate right skew'
    },
    'Pregnancies': {
        'Closest distribution': 'Poisson',
        'Reason': 'Count data with right-skewed shape'
    }
}

distribution_df = pd.DataFrame(distribution_match).T
print(distribution_df)
```
