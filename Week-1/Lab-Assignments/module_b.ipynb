**Background: Day 1 lecture recap:**

Descriptive statistics summarise the key properties of a dataset without making inferences beyond the data itself. 

>There are two families of measures:
- **Central tendency**: identifies the most representative value:
  - *Mean* μ = (1/n) Σxᵢ —> the arithmetic average; sensitive to outliers.
  - *Median* —> the middle value when data is sorted; robust to outliers.
  - *Mode* —> the most frequent value; especially useful for categorical and discrete data.
- **Dispersion**: quantifies how spread out the values are:
  - *Variance* σ² = (1/n) Σ(xᵢ − μ)² —> average squared deviation from the mean.
  - *Standard deviation* σ = √σ² —> same unit as the data; easier to interpret.
  - *Range* = max − min simplest measure; highly sensitive to a single outlier.

- **Visualisation** turns these numbers into patterns: histograms reveal distribution shape, boxplots expose medians and outliers, and pairplots show relationships between pairs of features.
- A key practical rule: when a feature is heavily skewed, the *median* is a more honest summary of the 'typical' value than the mean, because the mean is pulled towards the tail.

# Task 4: Descriptive Statistics
Compute central tendency and dispersion for key features, grouped by diabetes status.

> **Goal:** Compute and interpret central tendency and dispersion for key features, grouped by diabetes status.

```python
# ── Task 4 Setup (do not modify) ──────────────────────────────────────────────

import pandas as pd
import numpy as np

# scipy.stats provides statistical functions beyond basic numpy:
# skew(), kurtosis(), ttest_ind(), pearsonr(), etc.
from scipy import stats

URL = 'https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv'
df = pd.read_csv(URL)
df['DiabetesStatus'] = df['Outcome'].map({0: 'Non-Diabetic', 1: 'Diabetic'})

# List of the 8 numeric feature columns — excludes 'Outcome' (categorical)
# and 'DiabetesStatus' (string label). Used throughout Task 4.
NUMERIC_COLS = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
                'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']
print('Ready')
```

```python
# Provided: compute grouped descriptive statistics for four key features

# We focus on these four because they span different clinical domains:
# Glucose (metabolic), BMI (anthropometric), BloodPressure (cardiovascular), Age (demographic).
features = ['Glucose', 'BMI', 'BloodPressure', 'Age']

# groupby('DiabetesStatus') splits the DataFrame into two subsets:
#   - 'Diabetic'     (Outcome == 1)
#   - 'Non-Diabetic' (Outcome == 0)
# .agg([...]) applies multiple aggregation functions to each feature in one call.
# The result is a multi-level column DataFrame: (feature, statistic).
grouped = df.groupby('DiabetesStatus')[features].agg(['mean', 'median', 'std', 'var']).round(2)
print('Descriptive statistics by diabetes status:')

# Displaying without print() renders the DataFrame as a styled HTML table in Jupyter.
grouped
```

```python
# TODO: Find the mode of 'Pregnancies' and 'Age'.

# Also compute the range (max - min) for Glucose and Insulin.

# Hint: .mode()[0] for mode, .max() - .min() for range

# YOUR CODE HERE
mode_preg=df['Pregnancies'].mode()
mode_age=df['Age'].mode()
range_gi=df['Glucose'].max()-df['Glucose'].min()
range_i=df['Insulin'].max()-df['Insulin'].min()
print(mode_preg,mode_age,range_gi,range_i)
```

```python
# TODO: For ALL numeric features, compute a summary table showing:
# mean, median, std, variance, skewness, range (max-min)
# Present as a clean DataFrame rounded to 2 decimal places.

# YOUR CODE HERE
summary =pd.DataFrame({
    'mean': df[NUMERIC_COLS].mean(),
    'median': df[NUMERIC_COLS].median(),
    'std': df[NUMERIC_COLS].std(),
    'variance': df[NUMERIC_COLS].var(),
    'skewness': df[NUMERIC_COLS].skew(),
    'range': df[NUMERIC_COLS].max() - df[NUMERIC_COLS].min()
}).round(2)
print(summary)
```

```python
# TODO: For the Insulin feature, demonstrate the effect of outliers on the mean vs median.
# Step 1: Compute the original mean and median of Insulin.
# Step 2: Create a copy of the Insulin series and set 5 random values to 900
#         (simulate extreme outliers). Use np.random.seed(42) before selecting indices.
# Step 3: Compute the new mean and median.
# Step 4: Print a comparison table: Statistic | Before | After | Change

np.random.seed(42)

# YOUR CODE HERE
mean_insulin=df['Insulin'].mean()
median_insulin=df['Insulin'].median()
s=df['Insulin'].copy()
s[np.random.choice(s.index, 5, replace=False)] = 900
new_mean=s.mean()
new_median=s.median()
compare = pd.DataFrame({
    'Statistic': ['Mean', 'Median'],
    'Before': [mean_insulin, median_insulin],
    'After': [new_mean, new_median],
    'Change': [new_mean - mean_insulin, new_median - median_insulin]}).round(2)
print(compare)
```

```python
# ── Task 5 Setup (do not modify) ──────────────────────────────────────────────

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Apply a consistent visual theme so all plots in this task share the same style.
# style='whitegrid' adds light horizontal grid lines for readability.
# palette='muted' uses desaturated colours that print well in black-and-white.
sns.set_theme(style='whitegrid', palette='muted')

# Set the default figure size for all matplotlib figures created in this task.
plt.rcParams['figure.figsize'] = (9, 5)

URL = 'https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv'
df = pd.read_csv(URL)
df['DiabetesStatus'] = df['Outcome'].map({0: 'Non-Diabetic', 1: 'Diabetic'})
print('Ready')
```

```python
# Provided: overlapping histogram of Glucose by group
fig, ax = plt.subplots()
for status, group in df.groupby('DiabetesStatus'):
    ax.hist(group['Glucose'], bins=25, alpha=0.6, label=status, edgecolor='white')
ax.set_xlabel('Glucose (mg/dL)')
ax.set_ylabel('Frequency')
ax.set_title('Distribution of Glucose by Diabetes Status')
ax.legend()
plt.tight_layout()
plt.show()
```
