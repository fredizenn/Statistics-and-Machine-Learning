## Round 1: Statistics (Week 1, Day 3 content)

### Module A: Sampling Methods & the Central Limit Theorem (Tasks 1–3)
**Lecture source:** *Week 1 Day 3 → Sampling, Sampling Challenges, Central Limit Theorem*

| Task | What you'll do | Key concept|
|---|---|---|
| 1. Sampling Methods | Compare **random**, **systematic**, and **stratified** sampling on mean Glucose | Random sampling → equal chance for everyone. Systematic → every k-th row. Stratified → proportional draw from each subgroup (handles class imbalance). *(Cluster sampling is concept-only — know the definition for quiz questions, no code.)* |
| 2. Sampling Bias & Sampling Distributions | Show a biased sample (Diabetics-only) overestimates the population mean; simulate sampling distributions at n=10/30/100 | A **sampling distribution** = distribution of the sample statistic across many repeated samples. Bigger n → smaller spread (smaller standard error) |
| 3. CLT Simulation | Take repeated samples of skewed `Insulin` data, show sample means become Normal as n grows | **CLT:** for large enough n (rule of thumb n ≥ 30), the distribution of the sample mean is approximately Normal — *regardless of the population's original shape* |

**Must-know formula:** Standard Error `SE = σ / √n` → quadrupling n halves SE.

**Common trap:** Don't confuse "the population is Normal" with "the *sampling distribution of the mean* is Normal." CLT is about the latter.

### Module B: Confidence Intervals & Hypothesis Testing (Tasks 4–6)
**Lecture source:** *Week 1 Day 3 → Confidence Intervals, Hypothesis Testing, p-values, t-test/ANOVA, Correlation*

| Task | What you'll do | Key concept|
|---|---|---|
| 4. Confidence Intervals | Build 95% CIs for mean Glucose and mean BMI (Diabetics only); simulate CI coverage over 200 repeats | A 95% CI is **not** "95% probability the true mean is inside." Correct version: *"if we repeated this 100 times, ~95 of the intervals would contain the true mean."* |
| 5. Hypothesis Testing | t-test (Glucose, BMI by Diabetes status), ANOVA (Glucose across 3 age groups) | H₀ = no effect, H₁ = effect exists. p < 0.05 → evidence against H₀ (does **not** prove H₁, does **not** measure importance, does **not** imply causation) |
| 6. Correlation | Pearson correlation matrix + `pearsonr` significance test + ranked correlation table vs Outcome | Pearson r ranges −1 to +1; measures **linear** relationship only; **correlation ≠ causation** |

**Must-know rule of thumb:** t-test = 2 groups. ANOVA = 3+ groups (don't run 3 separate t-tests — inflates false positives).

**Common trap:** A tiny p-value (e.g. 1e-22) tells you statistical significance, not practical/clinical importance, always look at the *effect size* too.

## Round 2: Machine Learning (Week 2, Days 1–2 content)

### Module C: ML Foundations, Data Preparation & Feature Engineering (Tasks 7–9)
**Lecture source:** *Week 2 Day 1 → Supervised/Unsupervised/RL, ML Workflow, Missing Data, Noise vs Outliers, Feature Engineering, Scaling*

| Task | What you'll do | Key concept|
|---|---|---|
| 7. Identify the ML Task | Classify the problem (supervised classification), check class balance, fill in the 10-step ML workflow table | **Supervised** = labelled data (X→y). **Unsupervised** = no labels, find patterns. **Reinforcement** = learn via rewards/penalties |
| 8. Missing Values, Noise & Outliers | Mean vs median imputation (impossible zeros → NaN → impute); detect outliers with the **IQR rule** | Mean is pulled by outliers/skew → median is more robust. IQR rule: outlier if `< Q1 − 1.5×IQR` or `> Q3 + 1.5×IQR`. **Noise** = small random errors on many points; **outliers** = rare extreme points |
| 9. Feature Engineering & Scaling | Build `Glucose_Insulin_Ratio`, `BMI_Age_Interaction`, `AgeGroup` bins; apply Standardization (z-score) and Normalization (min–max) | **Standardization:** mean=0, std=1, `z=(x−μ)/σ`. **Normalization:** rescales to [0,1], `x'=(x−min)/(max−min)`. Use scaling whenever features are on very different ranges (e.g. Insulin 0–846 vs Age 21–81) |

**Must-know workflow:** Problem Definition → Data Collection → Preprocessing → Feature Selection → Data Splitting → Model Training → Hyperparameter Tuning → Evaluation → Deployment.

**Common trap:** Don't fit a scaler on the whole dataset before splitting — fit only on training data (data leakage). This is tested again in Module D.

### Module D: Feature Selection, Regression & Model Evaluation (Tasks 10–12)
**Lecture source:** *Week 2 Day 2 → Filter/Wrapper/Embedded Feature Selection, Train/Test Split, K-Fold CV, Linear/Polynomial Regression, Residuals*

| Task | What you'll do | Key concept|
|---|---|---|
| 10. Feature Selection, Train/Test Split & K-Fold CV | Filter (Pearson, ANOVA F-test), Wrapper (RFE, Forward Selection), Embedded (Lasso); build official train/test split (80/20, `stratify=y`); **NEW:** compare to 5-Fold Cross-Validation | **Filter** = fast statistical test before training. **Wrapper** = retrain model repeatedly to test feature subsets. **Embedded** = selection happens during training (Lasso zeros coefficients, Ridge shrinks but never to zero). **K-Fold CV** averages performance over k train/test splits → more stable estimate than one split |
| 11. Linear Regression | Simple LR (Glucose→BMI), then Multiple LR (all 7 features→BMI); interpret coefficients | Coefficient = "for every 1-unit increase in X, y changes by β, holding other features constant" |
| 12. Model Evaluation & Residuals | MAE, MSE, RMSE, R²; Polynomial Regression (degree 2) vs Linear; residual plots | **MAE/RMSE** = error in the target's original units (lower=better). **R²** = fraction of variance explained (0–1, higher=better). Good residuals = randomly scattered around 0, no pattern, constant spread |

**Must-know split rule:** SPLIT FIRST → fit scaler/selector on train only → transform test. Never fit anything on test data.

**Common trap:** Lasso (L1) can zero out coefficients → automatic feature selection. Ridge (L2) shrinks toward zero but (almost) never reaches exactly zero → does not do feature selection.

## Notation/Formula Quick Reference

| Concept | Formula |
|---|---|
| Standard Error | `SE = σ / √n` |
| Standardization (z-score) | `z = (x − μ) / σ` |
| Normalization (min–max) | `x' = (x − min) / (max − min)` |
| Lasso cost | `(1/n)Σ(yᵢ−ŷᵢ)² + α·Σ\|βⱼ\|` |
| Ridge cost | `(1/n)Σ(yᵢ−ŷᵢ)² + α·Σβⱼ²` |
| MAE | `(1/n)Σ\|yᵢ−ŷᵢ\|` |
| RMSE | `√[(1/n)Σ(yᵢ−ŷᵢ)²]` |
| R² | `1 − (SS_residual / SS_total)` |
